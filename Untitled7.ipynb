```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import urllib.request

# Preprocessing i metryki
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve, auc, f1_score

# Modele klasyczne
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor

# Deep Learning (LSTM Autoenkoder)
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, RepeatVector, TimeDistributed, Input

import warnings
warnings.filterwarnings('ignore')

np.random.seed(42)
tf.random.set_seed(42)


url = "https://raw.githubusercontent.com/numenta/NAB/master/labels/combined_labels.json"
response = urllib.request.urlopen(url)
labels_data = json.loads(response.read())

raw_datasets = [
    'nyc_taxi.csv',
    'machine_temperature_system_failure.csv',
    'ambient_temperature_system_failure.csv'
]

labeled_datasets = []

for dataset in raw_datasets:
    try:
        df = pd.read_csv(dataset)
        df['timestamp'] = pd.to_datetime(df['timestamp'])
        df['label'] = 0 
        
        matching_key = None
        for key in labels_data.keys():
            if key.endswith(dataset):
                matching_key = key
                break
                
        if matching_key:
            anomaly_timestamps = labels_data[matching_key]
            anomaly_timestamps = pd.to_datetime(anomaly_timestamps)
            df.loc[df['timestamp'].isin(anomaly_timestamps), 'label'] = 1
            
            new_filename = f"{dataset.replace('.csv', '')}_labeled.csv"
            df.to_csv(new_filename, index=False)
            labeled_datasets.append(new_filename)
            print(f"Oznaczono {dataset} -> Znaleziono anomalii: {df['label'].sum()}. Zapisano: {new_filename}")
        else:
            print(f"Nie znaleziono etykiet dla {dataset}!")
    except FileNotFoundError:
        print(f" Brak pliku {dataset} na dysku.")


#Trenowanie modeli

def create_sequences(X, y, time_steps=30):
    Xs, ys = [], []
    for i in range(len(X) - time_steps):
        Xs.append(X.iloc[i:(i + time_steps)].values)
        ys.append(y.iloc[i + time_steps]) 
    return np.array(Xs), np.array(ys)

def evaluate_and_save(y_true, y_pred, y_prob, model_name, dataset_name, log_file="wyniki_nab_zbiorcze.txt"):
    precision, recall, _ = precision_recall_curve(y_true, y_prob)
    auprc = auc(recall, precision)
    f1 = f1_score(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred)
    
    report_text = f"Wyniki: {model_name} | Zbiór: {dataset_name}\n"
    report_text += f"AUPRC: {auprc:.4f}\n"
    report_text += f"F1-Score: {f1:.4f}\n"
    report_text += "\nTablica pomyłek:\n" + str(cm) + "\n"
    report_text += "\nRaport klasyfikacji:\n" + classification_report(y_true, y_pred) + "\n"
    report_text += "="*60 + "\n\n"
    
    print(report_text)
    with open(log_file, "a", encoding="utf-8") as f:
        f.write(report_text)
    
    plt.figure(figsize=(5,3))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges')
    plt.title(f'CM: {model_name}\n({dataset_name})', fontsize=10)
    plt.ylabel('Rzeczywista etykieta')
    plt.xlabel('Przewidziana etykieta')
    
    safe_dataset_name = dataset_name.replace(".csv", "")
    safe_model_name = model_name.replace(" ", "_").lower()
    img_filename = f"{safe_dataset_name}_{safe_model_name}_cm.png"
    plt.savefig(img_filename, bbox_inches='tight', dpi=300)
    plt.show()
    return auprc, f1

if os.path.exists("wyniki_nab_zbiorcze.txt"):
    os.remove("wyniki_nab_zbiorcze.txt")

TIME_STEPS = 30 

for dataset in labeled_datasets:
    print(f"\n\n{'#'*60}")
    print(f"{'#'*60}")
    
    df = pd.read_csv(dataset)
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    
    # Standaryzacja kolumny label, jeśli trzeba
    if 'is_anomaly' in df.columns:
        df.rename(columns={'is_anomaly': 'label'}, inplace=True)
    elif 'anomaly' in df.columns:
        df.rename(columns={'anomaly': 'label'}, inplace=True)
        

    #Eksploracyjna Analiza (EDA) szeregu

    plt.figure(figsize=(15, 5))
    plt.plot(df['timestamp'], df['value'], color='blue', linewidth=1, label='Sygnał systemu', zorder=1)
    
    anomalies = df[df['label'] == 1]
    plt.scatter(anomalies['timestamp'], anomalies['value'], color='red', s=60, label='Prawdziwa anomalia', zorder=5)
    
    plt.title(f'Surowy sygnał systemu z oznaczonymi anomaliami\n({dataset})')
    plt.xlabel('Czas')
    plt.ylabel('Wartość sygnału')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.savefig(f'{dataset.replace(".csv", "")}_EDA_plot.png', bbox_inches='tight')
    plt.show()
    
    # Skalowanie i tworzenie okien czasowych
    scaler = StandardScaler()
    df['scaled_value'] = scaler.fit_transform(df[['value']])
    X_seq, y_seq = create_sequences(df[['scaled_value']], df['label'], TIME_STEPS)
    
    # Podział chronologiczny 70/30
    train_size = int(len(X_seq) * 0.7)
    X_train, X_test = X_seq[:train_size], X_seq[train_size:]
    y_train, y_test = y_seq[:train_size], y_seq[train_size:]
    
    X_train_flat = X_train.reshape(X_train.shape[0], X_train.shape[1] * X_train.shape[2])
    X_test_flat = X_test.reshape(X_test.shape[0], X_test.shape[1] * X_test.shape[2])
    

    # ISOLATION FOREST
    iforest = IsolationForest(n_estimators=100, contamination=0.05, random_state=42)
    iforest.fit(X_train_flat)
    if_preds = iforest.predict(X_test_flat)
    if_preds = np.where(if_preds == 1, 0, 1) 
    if_scores = -iforest.decision_function(X_test_flat) 
    evaluate_and_save(y_test, if_preds, if_scores, "Isolation Forest", dataset)
    

    # LOCAL OUTLIER FACTOR
    lof = LocalOutlierFactor(n_neighbors=20, novelty=True, contamination=0.05)
    lof.fit(X_train_flat)
    lof_preds = lof.predict(X_test_flat)
    lof_preds = np.where(lof_preds == 1, 0, 1)
    lof_scores = -lof.decision_function(X_test_flat)
    evaluate_and_save(y_test, lof_preds, lof_scores, "Local Outlier Factor", dataset)
    

    # LSTM AUTOENCODER
    X_train_normal = X_train[y_train == 0]
    
    model = Sequential([
        Input(shape=(X_train.shape[1], X_train.shape[2])),
        LSTM(16, activation='relu', return_sequences=False),
        RepeatVector(X_train.shape[1]),
        LSTM(16, activation='relu', return_sequences=True),
        TimeDistributed(Dense(X_train.shape[2]))
    ])
    
    model.compile(optimizer='adam', loss='mse')
    
    history = model.fit(X_train_normal, X_train_normal, epochs=15, batch_size=64, validation_split=0.1, verbose=1)

    plt.figure(figsize=(8, 5))
    plt.plot(history.history['loss'], label='Błąd treningowy (Train Loss)', color='green')
    plt.plot(history.history['val_loss'], label='Błąd walidacyjny (Val Loss)', color='red')
    plt.title(f'Krzywa uczenia LSTM Autoencodera\n({dataset})')
    plt.ylabel('Funkcja straty (MSE)')
    plt.xlabel('Epoka (Epoch)')
    plt.legend()
    plt.grid(True)
    plt.savefig(f'{dataset.replace(".csv", "")}_LSTM_learning_curve.png', bbox_inches='tight')
    plt.show()
    
    # Predykcja
    X_test_pred = model.predict(X_test, verbose=0)
    mse_test = np.mean(np.abs(X_test_pred - X_test), axis=(1, 2))
    
    X_train_pred = model.predict(X_train_normal, verbose=0)
    mse_train = np.mean(np.abs(X_train_pred - X_train_normal), axis=(1, 2))
    threshold = np.percentile(mse_train, 99.0)
    
    lstm_preds = np.where(mse_test > threshold, 1, 0)
    evaluate_and_save(y_test, lstm_preds, mse_test, "LSTM Autoencoder", dataset)
    
```
